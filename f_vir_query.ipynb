```python
import numpy as np
from scipy.special import erf
import matplotlib.pyplot as plt
from matplotlib import rcParams
import matplotlib.ticker
from PRaccretion import*
import math
plt.rcParams.update({
  "text.usetex": True,
  "font.family": "serif"
})


# Using v_L = V_NL (v_P)


def v_pbh_linear_regime(z): # in units of pc/yr
    return  v_L_Serpico(z)


# Function to calculate M_2_sigma as a function of redshift
def M2_sigma(z): #in units of solar mass
    return 1e8 * ((v_pbh_linear_regime(z)/(17 * 1e3 * (yr/pc)))**3) *(((z+1)/10)**(-3/2)) #8.8e12 * np.exp(-1.8 * (z + 1))









# Function to calculate M_2_sigma as a function of redshift
def M2_sigma(z): #in units of solar mass
    return 1e8 * ((v_pbh_linear_regime(z)/(17 * 1e3 * (yr/pc)))**3) *(((z+1)/10)**(-3/2)) 

def sigma(z):
        return  10.2 - 0.79 * np.log10(M2_sigma(z))
def f_vir(z):
    def D(z):
        return (1+z)
    def nu_min(z):
        return (delta_c * D(z))/sigma(z)
   
    return 1 - erf(nu_min(z) / np.sqrt(2))

```

```python
#delta = delta_c * D(z) = delta_c * (1+z)
delta_c =  1.686  # Critical density for collapse, dimensionless







fig,ax = plt.subplots(1, figsize = (9, 7))
plt.rc('lines', linewidth = 3.65)
plt.rcParams['font.family'] = 'serif'
plt.tick_params(axis = 'both', right=True, top=True, which = 'minor', width= 1.2, length = 4, direction = 'in')
plt.tick_params(axis = 'both', right=True, top=True, which = 'major', width= 1.2, length = 9, direction = 'in')

# Redshift range
z = np.geomspace(1, 1e4, 10000)  

ax.tick_params(axis='x', pad = 7.9)
ax.tick_params(axis='y', pad = 7.9)

plt.xticks(fontsize = 33)
plt.yticks(fontsize = 33)

plt.ylim(1e-3, 2)



plt.loglog(z, f_vir(z), color='#04D8B2')
plt.xlabel('$(1+z)$',  fontsize = 28)
plt.ylabel('$f_{\mathrm{pbh},\mathrm{vir}}$', fontsize = 38.1, labelpad = 28)

plt.text(1.6e3, 0.5, "$ \delta_c = 1.686$", fontsize=26, ha='center', va='center')
plt.text(2e3, 0.25, "$ \delta = \delta_c(1+z)$", fontsize=26, ha='center', va='center')


plt.title(r'$v_\mathrm{pbh}>v_\mathrm{pbh,L}$', fontsize = 35.7, pad = 9)


plt.savefig('f_vir1.pdf', bbox_inches = 'tight')
plt.show()
```
